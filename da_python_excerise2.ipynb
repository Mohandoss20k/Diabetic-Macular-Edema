Create at least 3 charts using Matplotlib


```python
import matplotlib.pyplot as plt
```

Top 10 customers by total cost


```python
top10=df.groupby("Customer_Name")["Cost"].sum()
top10=top10.sort_values(ascending=False)
```

```python
top10.head()
```

```python
top10=top10.head(10)
top10
```

```python
top10.plot(kind='bar',figsize=(10,5))
plt.title("Top 10 Customers by Total Cost")
plt.xlabel("Customer Name")
plt.ylabel("Total Cost")
plt.show()
```

Cost distribution

```python
plt.figure(figsize=(5,8))

plt.hist(df["Cost"].dropna() ,bins=20)
plt.title("Distribution of Shipment Costs")
plt.xlabel("cost")
plt.ylabel("Number of shipment")
plt.tight_layout()
plt.show()
```

Total delay by route

```python
route_delay = (
    df.groupby("Route")["Delay_Days"]
    .sum()
    .sort_values(ascending=False)
)

route_delay.plot(kind="bar", figsize=(10, 5))

plt.title("Total Delay Days by Route")
plt.xlabel("Route")
plt.ylabel("Total Delay Days")
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```
